```python
import torch

# 1. Check if CUDA (GPU) is available
print(torch.cuda.is_available())  # Returns True or False

# 2. Get the number of available CUDA devices
print(torch.cuda.device_count())

# 3. Get the name of the active GPU
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
    print(torch.version.cuda)  # Check the CUDA version PyTorch was built with
```
